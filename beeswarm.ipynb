```python
# Let's generate a simple SHAP-like beeswarm plot for healthcare expense prediction

import matplotlib.pyplot as plt
import numpy as np

# Mock feature names
features = ['age', 'male', 'income', 'smoker_status', 'chronic_conditions']

# Create synthetic SHAP values for each feature
np.random.seed(42)
shap_values = [
    np.random.normal(loc=0.5, scale=0.2, size=50),   # age: high positive effect
    np.random.normal(loc=-0.2, scale=0.1, size=50),  # male: slight negative
    np.random.normal(loc=0.1, scale=0.3, size=50),   # income: mixed
    np.random.normal(loc=0.6, scale=0.25, size=50),  # smoker_status: strong positive
    np.random.normal(loc=0.0, scale=0.15, size=50),  # chronic_conditions: centered
]

# Assign random feature values (for color coding)
feature_values = [
    np.random.uniform(20, 90, size=50),  # age
    np.random.randint(0, 2, size=50),    # male (0 or 1)
    np.random.uniform(30000, 120000, size=50),  # income
    np.random.randint(0, 2, size=50),    # smoker_status (0 or 1)
    np.random.randint(0, 5, size=50),    # number of chronic conditions
]

# Plot setup
fig, ax = plt.subplots(figsize=(8, 5))

for i, (shap_feat, feat_val) in enumerate(zip(shap_values, feature_values)):
    y = np.ones_like(shap_feat) * i
    ax.scatter(shap_feat, y,
               c=feat_val,
               cmap='coolwarm',
               edgecolor='k',
               alpha=0.7)

# Formatting
ax.set_yticks(range(len(features)))
ax.set_yticklabels(features)
ax.set_xlabel("SHAP Value (impact on model output)")
ax.set_title("SHAP Summary (Beeswarm) Plot Example")
plt.gca().invert_yaxis()  # most important feature at top
plt.tight_layout()

# Save
beeswarm_output = "/mnt/data/beeswarm_example.png"
plt.savefig(beeswarm_output)

beeswarm_output

```
