In this notebook, you'll get hands-on practice using SQL to explore synthetic EHR data from Synthea.

```python
# Specify the disk drive path to the folder containing the Synthea CSV files
# You'll change the path on the next line to match your system setup
my_path = "/Users/spb24/Downloads/coherent-11-07-2022/csv/"
```

```python
import pandas
import sqlite3

# Load patient data from CSV file
patients = pandas.read_csv(my_path + "patients.csv")
```

```python
# Create in-memory SQLite database
rdbms = sqlite3.connect(":memory:")

# Store the patients DataFrame as a SQL table
patients.to_sql("patients", rdbms, index=False, if_exists="replace")
```

### 🔹 Select the First Few Rows

- `SELECT * FROM patients` returns all columns
- `LIMIT 5` restricts output to 5 rows

```python
query = """
SELECT * FROM patients
LIMIT 5;
"""

pandas.read_sql_query(query, rdbms)
```

✏️ **Try it yourself**  
Modify the SQL query below to return only the columns `Id`, `FIRST`, and `BIRTHDATE`.

```python
query = """
SELECT Id, FIRST, BIRTHDATE
FROM patients
LIMIT 5;
"""

pandas.read_sql_query(query, rdbms)
```

### 🔹 Filtering Rows with WHERE

- `WHERE GENDER = 'M'` restricts the output to male patients
- You can use logical operators like `=`, `IN`, `>`, `LIKE`, etc.

```python
query = """
SELECT Id, FIRST, LAST, GENDER
FROM patients
WHERE GENDER = 'M'
LIMIT 10;
"""

pandas.read_sql_query(query, rdbms)
```

### 🔹 Counting Records with GROUP BY

- `GROUP BY GENDER` groups all records by gender
- `COUNT(*)` returns the number of patients in each group

```python
query = """
SELECT GENDER, COUNT(*) AS num_patients
FROM patients
GROUP BY GENDER;
"""

pandas.read_sql_query(query, rdbms)
```
